```python
num = int(input('Enter a number:')) #declared as global var can access outside the fun
def OddEven():    
    if(num % 2 == 0):
        return 'even'
    else:
        return 'odd'
result=OddEven()
if (result == 'even'):
    print(f'{num} is Even number')
else:
    print(f'{num} is Odd number')
```

```python
gen = str(input('Your Gender:'))
age = int(input('Your Age:'))

def Elegible():
    if(gen == 'Male' and age > 21):
        return 'ElegiblityForMarriage'
    elif(gen == 'Female' and age > 18):
        return 'ElegiblityForMarriage'
    else:
        return 'NOT ELIGIBLE'
result = Elegible()
print(result)
```

```python
s1=int(input('Subject1='))
s2=int(input('Subject2='))
s3=int(input('Subject3='))
s4=int(input('Subject4='))
s5=int(input('Subject5='))
def percentage():
    add=s1+s2+s3+s4+s5
    return add
result = percentage()
per=float(result/5)
print('Total:',result)
print('Percentage:',per)
```

```python
def triangle(h,b):
    print('Height:',h)
    print('Breadth:',b)
    return (h*b)/2
triangle =triangle(32,34)
print('Area of Triangle:',triangle)
def perimeter(h1,h2,b):
    print('Height1:',h1)
    print('Height2:',h2)
    print('Breadth:',b)
    return (h1+h2+b) 
perimeter = perimeter(2,4,4)
print('Area of perimeter:',perimeter)
```
