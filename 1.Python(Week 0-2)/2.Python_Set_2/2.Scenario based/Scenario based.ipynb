```python
user_id=int(input())
count=0
for i in range(1,user_id+1):
    if user_id%i==0:
        count=count+1
if count == 2:
    print(f'{user_id} is prime so, user can login')
else:
    print(f'{user_id} is not prime so, user not able login')
```
